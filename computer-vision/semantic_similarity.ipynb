```python
from PIL import Image
```

```python
dog1 = Image.open('/content/beagle-hound-dog.jpg')
dog2 = Image.open('/content/gettyimages-586890581.jpg')
cat1 = Image.open('/content/british-shorthair-8032816_640.jpg')
```

```python
import torchvision
```

```python
torchvision.models.list_models()
```

```python
weights = torchvision.models.EfficientNet_B0_Weights.IMAGENET1K_V1
```

```python
model = torchvision.models.efficientnet_b0(weights=weights)
```

```python
[name for name, params in model.named_children()]
```

```python
model.avgpool
```

```python
layers = list(model.children())[:-1]
```

```python
import torch.nn as nn
```

```python
model = nn.Sequential(*layers)
```

```python
model
```

```python
transform = weights.transforms()
processed_dog1 = transform(dog1).view(1, 3, 224, 224)
processed_dog2 = transform(dog2).view(1, 3, 224, 224)
processed_cat1 = transform(cat1).view(1, 3, 224, 224)
```

```python
processed_dog1.shape
```

```python
dog_embed1 = model(processed_dog1)
dog_embed2 = model(processed_dog2)
cat_embed1 = model(processed_cat1)
```

```python
vectorized_dog1 = dog_embed1.squeeze()
vectorized_dog2 = dog_embed2.squeeze()
vectorized_cat1 = cat_embed1.squeeze()
```

```python
import torch.nn.functional as F
```

```python
dog1dog2 = F.cosine_similarity(vectorized_dog1, vectorized_dog2, dim=0).item()
dog1cat1 = F.cosine_similarity(vectorized_dog1, vectorized_cat1, dim=0).item()
dog2cat1 = F.cosine_similarity(vectorized_dog2, vectorized_cat1, dim=0).item()
```

```python
print(f'Dog1 vs Dog2: {dog1dog2}\nDog1 vs Cat1: {dog1cat1}\nDog2 vs Cat1: {dog2cat1}')
```

```python
model = torchvision.models.efficientnet_b0(weights=weights)
```

```python
model.classifier[1] = nn.Linear(1280, 1)
```

```python
model.classifier
```

```python
for parameter in model.parameters():
    parameter.requires_grad_ = False
```

```python
for parameter in model.parameters():
    parameter.requires_grad_ = True
```

```python
from torchvision.datasets import ImageFolder
from torch.utils.data import DataLoader
```

```python
full_dataset = ImageFolder(
    '/content/PetImages',
    transform=transform,
)
```

```python
train_loader = DataLoader(full_dataset, batch_size=32, shuffle=True)
```

```python
def train(model, optimizer, criterion, train_loader, n_epochs, device):

  for epoch in range(n_epochs):
    total_loss = 0.
    for X_batch, y_batch in train_loader:
      X_batch, y_batch = X_batch.to(device), y_batch.to(device)
      preds = model(X_batch)
      loss = criterion(preds, y_batch.unsqueeze(1).float())
      total_loss += loss.item()
      loss.backward()
      optimizer.step()
      optimizer.zero_grad()
    print(f'Epoch {epoch+1}/{n_epochs}, Loss: {total_loss / len(train_loader)}')
```

```python
import torch
```

```python
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
criterion = nn.BCEWithLogitsLoss()
```

```python
if torch.cuda.is_available():
    device='cuda'
elif torch.backends.mps.is_available():
    device='mps'
else:
    device='cpu'
```

```python
model.to(device)
```

```python
train(model, optimizer, criterion, train_loader, 5, device)
```

```python
layers = list(model.children())[:-1]
model = nn.Sequential(*layers)
```

```python
dog_embed1 = model(processed_dog1.to(device)).squeeze()
dog_embed2 = model(processed_dog2.to(device)).squeeze()
cat_embed1 = model(processed_cat1.to(device)).squeeze()
```

```python
dog1dog2 = F.cosine_similarity(dog_embed1, dog_embed2, dim=0).item()
dog1cat1 = F.cosine_similarity(dog_embed1, cat_embed1, dim=0).item()
dog2cat1 = F.cosine_similarity(dog_embed2, cat_embed1, dim=0).item()
```

```python
print(f'Dog1 vs Dog2: {dog1dog2}\nDog1 vs Cat1: {dog1cat1}\nDog2 vs Cat1: {dog2cat1}')
```
